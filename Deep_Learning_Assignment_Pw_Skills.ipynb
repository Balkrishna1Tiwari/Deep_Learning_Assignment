### 1 What is TensorFlow 2.0, and how is it different from TensorFlow 1.x?

TensorFlow 2.0 is a major upgrade over TensorFlow 1.x. It simplifies APIs, uses eager execution by default, and tightly integrates with the Keras API.

**Key Differences**:
- Eager execution by default
- `tf.function` for graph execution
- Unified high-level Keras API
- Cleaner and more Pythonic

### 5 How do you create a neural network using TensorFlow 2.0?

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(64, activation='relu', input_shape=(784,)),
    Dense(10, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
```

### 12 What is the basic structure of a PyTorch neural network?

```python
import torch.nn as nn

class Net(nn.Module):
    def __init__(self):
        super(Net, self).__init__()
        self.fc1 = nn.Linear(784, 64)
        self.fc2 = nn.Linear(64, 10)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        return self.fc2(x)
```
